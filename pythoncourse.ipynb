### 1. Numeric Types 
* int: Holds integers of unlimited length.
* float: Holds accurate floating-point numbers up to 15 decimal places.
* complex: Written with a j or J suffix to denote the imaginary component.

### 2. Sequence Types
* str: Written inside single ('), double ("), or triple (""" for multi-line) quotes. Strings are immutable.
* list: Declared using square brackets []. You can modify, append, or delete elements after creation.
* tuple: Declared using parentheses (). They look like lists but protect data from accidental modification. Note: A single-element tuple requires a trailing comma (e.g., (5,)).

### 3. Mapping Type
* dict: Declared using curly braces {} with key: value pairs. Keys must be unique and immutable

### 4. Set Types
* set: Declared using curly braces {} containing standalone values. Automatically filters out duplicate entries.

### 5. Boolean Type
* bool: Logical values restricted to either True or False

### 6. None Type
* NoneType: Represented by the keyword None, signifying the absence of a value or a null object.

## __OPERATORS__
* Python operators are special symbols or keywords used to perform computations on variables and values. They are the foundational building blocks of expression logic in programming.

__Python organizes its operators into seven core functional categories:__
* __Arithmetic Operators:__ Perform mathematical calculations like +, -, *, / (float division), // (floor division), % (modulus), and ** (exponentiation).
* __Assignment Operators:__ Assign values to variables, including combined operators like =, +=, -=, *=, /=, %=, //=, and **=.
* __Comparison Operators:__ Compare values, returning Boolean True/False using ==, !=, >, <, >=, and <=.
* __Logical Operators:__ Combine conditional statements using and, or, and not.
* __Identity Operators:__ Compare object memory locations with is and is not.
* __Membership Operators:__ Check if a value is present in a sequence using in and not in.
* __Bitwise Operators:__ Perform bit-level operations on integers with &, |, ^, ~, <<, and >>.

```python
# 1. ARITHMETIC OPERATORS
# ==============================================================================
print("--- 1. Arithmetic Operators ---")
a, b = 10, 3

print(f"Addition (a + b): {a + b}")
print(f"Subtraction (a - b): {a - b}")
print(f"Multiplication (a * b): {a * b}")
print(f"Float Division (a / b): {a / b}")      # Returns a float
print(f"Floor Division (a // b): {a // b}")    # Truncates fractional part
print(f"Modulus (a % b): {a % b}")            # Returns the remainder
print(f"Exponentiation (a ** b): {a ** b}")    # 10 to the power of 3
```

```python
# 2. ASSIGNMENT OPERATORS
# ==============================================================================
print("\n--- 2. Assignment Operators ---")
x = 5
print(f"Initial assignment (x = 5): {x}")

x += 3
print(f"Add and assign (x += 3): {x}")

x -= 2
print(f"Subtract and assign (x -= 2): {x}")

x *= 2
print(f"Multiply and assign (x *= 2): {x}")

x /= 4
print(f"Divide and assign (x /= 4): {x}")
```

```python
# 3. COMPARISON OPERATORS
# ==============================================================================
print("\n--- 3. Comparison Operators ---")
print(f"Is a equal to b? (a == b): {a == b}")
print(f"Is a not equal to b? (a != b): {a != b}")
print(f"Is a greater than b? (a > b): {a > b}")
print(f"Is a less than b? (a < b): {a < b}")
print(f"Is a greater than or equal to b? (a >= b): {a >= b}")
print(f"Is a less than or equal to b? (a <= b): {a <= b}")
```

```python
# 4. LOGICAL OPERATORS
# ==============================================================================
print("\n--- 4. Logical Operators ---")
cond1 = (a > b)   # True (10 > 3)
cond2 = (b == 5)  # False (3 == 5)

print(f"Logical AND (cond1 and cond2): {cond1 and cond2}")  # Both must be True
print(f"Logical OR (cond1 or cond2): {cond1 or cond2}")    # At least one must be True
print(f"Logical NOT (not cond1): {not cond1}")              # Inverts the value
```

```python
# 5. IDENTITY OPERATORS
# ==============================================================================
print("\n--- 5. Identity Operators ---")
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1  # Points to the exact same memory location as list1

print(f"Do list1 and list2 reference the same object? (list1 is list2): {list1 is list2}")
print(f"Do list1 and list3 reference the same object? (list1 is list3): {list1 is list3}")
print(f"Are list1 and list2 distinct objects? (list1 is not list2): {list1 is not list2}")
```

```python
# 6. MEMBERSHIP OPERATORS
# ==============================================================================
print("\n--- 6. Membership Operators ---")
fruits = ["apple", "banana", "cherry"]

print(f"Is 'banana' in the fruits list? ('banana' in fruits): {'banana' in fruits}")
print(f"Is 'orange' missing from the fruits list? ('orange' not in fruits): {'orange' not in fruits}")
```

```python
# 7. BITWISE OPERATORS
# ==============================================================================
print("\n--- 7. Bitwise Operators ---")
# m = 5 (0101 in binary)
# n = 3 (0011 in binary)
m, n = 5, 3

print(f"Bitwise AND (m & n): {m & n}")    # 0101 & 0011 = 0001 (1)
print(f"Bitwise OR (m | n): {m | n}")     # 0101 | 0011 = 0111 (7)
print(f"Bitwise XOR (m ^ n): {m ^ n}")    # 0101 ^ 0011 = 0110 (6)
print(f"Bitwise NOT (~m): {~m}")          # Inverts bits: -(m + 1) -> -6
print(f"Left Shift (m << 1): {m << 1}")   # Shifts bits left (multiplies by 2) -> 1010 (10)
print(f"Right Shift (m >> 1): {m >> 1}")  # Shifts bits right (floor divides by 2) -> 0010 (2)
```

## Pythonic Style Conventions (Best Practices)
The Python community follows official style guidelines known as PEP 8. Following these ensures your code looks professional and is easy for others to read. 
* Use Snake Case: Write variable names in lowercase, joining multi-word names with underscores (e.g., user_account_balance).
* Be descriptive: Choose names that explain what the data holds. Use elapsed_time_seconds instead of just t.
* Constants in Uppercase: If a variable is meant to stay the same throughout your program, write it in all caps (e.g., MAX_LOGIN_ATTEMPTS = 5).
* Avoid confusing characters: Try not to use lowercase l (L), uppercase O, or uppercase I as single-letter names, as they look like numbers 1 and 0.
