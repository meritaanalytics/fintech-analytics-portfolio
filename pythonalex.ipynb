```python
# Outer list: Days of the week
days = ["Monday", "Tuesday"]

# Inner list: Time shifts
shifts = ["Morning", "Evening"]

# Outer loop tracks each day
for day in days:
    print(f"📅 Schedule for {day}:")
    
    # Inner loop tracks shifts within that specific day
    for shift in shifts:
        print(f"   ↳ 🕒 {shift} Shift Cover")
        
    print("-" * 25)  # Visual separator after inner loop finishes
```

# BMI Calculator

```python
weight  =float(input("enter the weight in kgs"))
print("Your weight: ", weight)
height = float(input(" enter the height  in meters"))
print("Your height: ", height)
BMI = weight / height**2
print("your BMI IS: ", BMI)
if  BMI > 0:
    if BMI <= 18.5:
        print("you are underweight")
    elif BMI <= 25:
        print("normal weight")
    elif BMI <= 30:
        print("overweight")
    else:
        print("severe overweight")
    
```

```python
def calculate_bmi():
    print("--- Welcome to the BMI Calculator ---")
    
    try:
        # 1. Capture and convert user input
        weight = float(input("Enter your weight in kilograms (e.g., 70): "))
        height = float(input("Enter your height in meters (e.g., 1.75): "))
        
        # Validate inputs are above zero
        if weight <= 0 or height <= 0:
            print("Weight and height must be greater than zero.")
            return

        # 2. Calculate BMI (weight divided by height squared)
        bmi = weight / (height ** 2)
        
        # 3. Round the result for readability
        bmi = round(bmi, 2)
        print(f"\nYour BMI is: {bmi}")
        
        # 4. Determine health category based on standard ranges
        if bmi < 18.5:
            category = "Underweight"
        elif 18.5 <= bmi < 25:
            category = "Normal weight (Healthy)"
        elif 25 <= bmi < 30:
            category = "Overweight"
        else:
            category = "Obese"
            
        print(f"Health Category: {category}")
        
    except ValueError:
        print("Invalid input! Please enter numbers only.")

# Run the calculator
```

# AUTOMATIC FILE SORTER IN FILE EXPLORER

```python
import os, shutil
```

```python
path = "D:\\"
```

```python
os.listdir(path)
```

```python
file_name = os.listdir(path)
```

```python
folder_names = ['csv files', 'image files', 'text files', 'pdf files', 'excel files']
for loop in range (0,5):
    if not os.path.exists(path+ folder_names[loop]):
        os.makedirs(path+ folder_names[loop])
```

```python
for file in file_name:
    if '.csv' in file and not os.path.exists(path+ 'csv files/' + file ):
        shutil.move( path + file, path + 'csv files/' + file)
    elif '.webp' in file and not os.path.exists(path+ 'image files/' + file ):
        shutil.move( path + file, path + 'image files/' + file)
    elif '.txt' in file and not os.path.exists(path+ 'text files/' + file ):
        shutil.move( path + file, path + 'text files/' + file)
    elif '.pdf' in file and not os.path.exists(path+ 'pdf files/' + file ):
        shutil.move( path + file, path + 'pdf files/' + file)
    elif '.XLSX' in file and not os.path.exists(path+ 'excel files/' + file ):
        shutil.move( path + file, path + 'excel files/' + file)
    else:
          print("there are files in this path that were moved")
```
